```python
!pip -q install python-docx
import glob, pandas as pd
from docx import Document
from docx.shared import Pt
from google.colab import files

folders = ["results", "results_unsw", "results_v2_nsl", "results_v2_unsw",
           "results_v3b_nsl", "results_v3b_unsw", "logs", "logs_unsw"]
           doc = Document(); doc.add_heading("AFO experiment logs", 0)
           for fo in folders:
               for f in sorted(glob.glob(fo + "/**/*", recursive=True)):
                       if f.endswith(".csv"):
                                   df = pd.read_csv(f).head(60).round(4)
                                               doc.add_heading(f, 2)
                                                           t = doc.add_table(rows=1, cols=len(df.columns)); t.style = "Table Grid"
                                                                       for i, c in enumerate(df.columns): t.rows[0].cells[i].text = str(c)
                                                                                   for _, r in df.iterrows():
                                                                                                   cells = t.add_row().cells
                                                                                                                   for i, v in enumerate(r): cells[i].text = str(v)
                                                                                                                           elif f.endswith(".txt"):
                                                                                                                                       doc.add_heading(f, 2)
                                                                                                                                                   p = doc.add_paragraph(open(f).read()); p.runs[0].font.size = Pt(8)
                                                                                                                                                   doc.save("afo_logs.docx"); files.download("afo_logs.docx")
```
